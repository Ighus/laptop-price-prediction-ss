### First look inside the data

##### **Dataset Columns Description**
1. **`graphic card type`**  
   The type of graphics card in the laptop (e.g., integrated, NVIDIA, AMD). Influences the price and performance in tasks like gaming or graphic editing.

2. **`communications`**  
   Communication technologies available in the laptop (e.g., Wi-Fi, Bluetooth, Ethernet).

3. **`resolution (px)`**  
   Screen resolution of the laptop, specified in pixels (e.g., 1920x1080). Relates to display quality.

4. **`CPU cores`**  
   The number of cores in the processor. Affects performance in multitasking and intensive workloads.

5. **`RAM size`**  
   The amount of RAM in the laptop (e.g., 8GB, 16GB), a crucial factor for overall performance.

6. **`operating system`**  
   The operating system installed on the laptop (e.g., Windows 10, macOS, Linux). May affect the price depending on the licensing.

7. **`drive type`**  
   The type of storage in the laptop (e.g., HDD, SSD). SSDs offer faster performance and are usually more expensive.

8. **`input devices`**  
   Input devices included in the laptop (e.g., backlit keyboard, touchpad, fingerprint reader).

9. **`multimedia`**  
   Multimedia features available in the laptop (e.g., speakers, microphones, webcam).

10. **`RAM type`**  
    The type of RAM installed (e.g., DDR3, DDR4), impacting performance and compatibility.

11. **`CPU clock speed (GHz)`**  
    The clock speed of the processor, measured in GHz. Higher speeds generally indicate better performance for single-threaded tasks.

12. **`CPU model`**  
    The specific model of the processor (e.g., Intel Core i5-1135G7, AMD Ryzen 7 5800H). Provides detailed processor capabilities.

13. **`state`**  
    The condition of the laptop (e.g., new, refurbished, used). Directly impacts the price.

14. **`drive memory size (GB)`**  
    The storage capacity in GB (e.g., 256GB, 1TB). Larger capacities often come at a higher price.

15. **`warranty`**  
    The warranty period provided by the manufacturer (e.g., 1 year, 2 years).

16. **`screen size`**  
    The size of the screen, typically measured in inches (e.g., 14", 15.6"). Larger or higher-quality screens may increase the cost.

17. **`buynow_price`**  
    The buy-now price of the laptop. This is the target variable for the prediction model.

Checking the null values

Analysis of the 4711 entries reveals a substantial presence of null values within the CPU clock speed (GHz) and RAM type columns, both of which are critical for the analytic process. Consequently, a data imputation strategy will be employed. For columns exhibiting nullity below 10%, mode or median imputation will be utilized, with subsequent impact assessment. An alternative methodology involves the creation of an 'unknown' categorical variable.

```python
null_summary = pd.DataFrame({
  "Column": df.columns,
  "Null count": df.isnull().sum(),
  "Percentage Null": ((df.isnull().sum() / len(df)) * 100).round(2)
}).sort_values(by="Null count", ascending=False)

fig = px.bar(null_summary, x="Column", y="Null count",
             title="Null Values by Column", text="Percentage Null",template="plotly_dark")
fig.update_layout(xaxis_title="Columns", yaxis_title="Count of Nulls")
fig.show()
```

```python
def impute_data(df, mode_cols=None, median_cols=None, unknown_cols=None,
                auto=True):
    """
    Imputa valores ausentes.
    · Se auto=True e todas as listas forem None/vazias, decide:
        numéricas  -> mediana
        categóricas-> moda
    """
    # ------------------ detectar tipos ------------------
    if auto and not any([mode_cols, median_cols, unknown_cols]):
        median_cols  = df.select_dtypes(include='number').columns.tolist()
        mode_cols    = df.select_dtypes(exclude='number').columns.tolist()
        unknown_cols = []                     # opcional

    mode_cols, median_cols, unknown_cols = map(
        lambda x: [] if x is None else x,
        [mode_cols, median_cols, unknown_cols])

    all_cols = mode_cols + median_cols + unknown_cols
    if not all_cols:
        raise ValueError("Nenhuma coluna para imputar.")

    # ------------------ construir mapa ------------------
    impute_map = {}
    for c in mode_cols:
        m = df[c].mode(dropna=True)
        impute_map[c] = m.iat[0] if not m.empty else np.nan
    for c in median_cols:
        impute_map[c] = df[c].median()
    for c in unknown_cols:
        impute_map[c] = "unknown"

    return df.fillna(impute_map)

df_validated = impute_data(df, 
                           mode_cols=['operating system', 'resolution (px)', 'drive type', 'drive memory size (GB)', 'screen size', 'RAM size'], 
                           median_cols=[], 
                           unknown_cols=['RAM type', 'communications', 'input devices', 'CPU model', 'graphic card type'])
```

```python
# Função para visualização automatizada de comparação antes e depois da imputação
def comparing_data(df_original, df_imputado, cols_numericas, cols_categoricas):
    for col in cols_numericas:
        fig = go.Figure()

        fig.add_trace(go.Histogram(
            x=df_original[col].dropna(),
            name=f'{col} (Antes)',
            opacity=0.75
        ))

        fig.add_trace(go.Histogram(
            x=df_imputado[col],
            name=f'{col} (Depois)',
            opacity=0.75
        ))

        fig.update_layout(
            title=f'Comparação da Distribuição - {col}',
            xaxis_title=col,
            yaxis_title='Contagem',
            barmode='overlay',
            template='plotly_dark'
        )

        fig.show()

    for col in cols_categoricas:
        original_count = df_original[col].value_counts().reset_index()
        imputado_count = df_imputado[col].value_counts().reset_index()

        original_count.columns = [col, 'Contagem']
        imputado_count.columns = [col, 'Contagem']

        fig = go.Figure()

        fig.add_trace(go.Bar(
            x=original_count[col],
            y=original_count['Contagem'],
            name=f'{col} (Antes)'
        ))

        fig.add_trace(go.Bar(
            x=imputado_count[col],
            y=imputado_count['Contagem'],
            name=f'{col} (Depois)'
        ))

        fig.update_layout(
            title=f'Comparação da Distribuição Categórica - {col}',
            xaxis_title=col,
            yaxis_title='Contagem',
            barmode='group',
            template='plotly_dark'
        )

        fig.show()

cols_numericas = []
cols_categoricas = ['operating system', 'resolution (px)', 'drive type', 'drive memory size (GB)', 'screen size', 'RAM size', 'RAM type', 'communications', 'input devices', 'CPU model', 'graphic card type']
comparing_data(df, df_validated, cols_numericas, cols_categoricas)
```

```python
def plot_correlacao_pre_pos(df_original, df_imputado):
    """Gera heatmaps lado a lado e coloca a barra de cor fora dos gráficos."""
    corr_before = df_original.corr(numeric_only=True)
    corr_after  = df_imputado.corr(numeric_only=True)

    fig = make_subplots(rows=1, cols=2,
                        subplot_titles=(
                            'Correlação Antes da Imputação',
                            'Correlação Depois da Imputação'),
                        horizontal_spacing=0.07)

    # Heatmap before
    fig.add_trace(go.Heatmap(
        z=corr_before.values,
        x=corr_before.columns,
        y=corr_before.columns,
        colorscale='RdBu', zmin=-1, zmax=1,
        showscale=False),
        row=1, col=1)

    # Heatmap after
    fig.add_trace(go.Heatmap(
        z=corr_after.values,
        x=corr_after.columns,
        y=corr_after.columns,
        colorscale='RdBu', zmin=-1, zmax=1,
        colorbar=dict(title='Correlação', x=1.02)),
        row=1, col=2)

    fig.update_layout(template='plotly_dark', height=600, width=1200,
                      margin=dict(l=60, r=60, t=60, b=60))
    fig.show()

plot_correlacao_pre_pos(df, df_validated)
```

### Modeling data

```python
def to_list(value):
    if pd.isnull(value):
        return []

    try:
        return ast.literal_eval(value)
    except:
        return []

# 1. Applying the function to the 'communications' column
df['communications_list'] = df['communications'].apply(to_list)

# 2. Creating a set to store all unique items
all_itens = set()

# 3. Adding itens from each list to the set
for lista in df['communications_list']:
    for item in lista:
        all_itens.add(item.strip().lower())

# 4. Printing itens to create the mapping
print(all_itens)

mapping_comunications = {
    'wi-fi': ['wi-fi'],
    'wi-fi_bgnac': ['wi-fi 802.11 b/g/n/ac'],
    'wi-fi_bgn': ['wi-fi 802.11 b/g/n'],
    'wi-fi_abgn': ['wi-fi 802.11 a/b/g/n'],
    'wi-fi_abgnac': ['wi-fi 802.11 a/b/g/n/ac'],
    'bluetooth': ['bluetooth'],
    'lan_100': ['lan 10/100 mbps'],
    'lan_1000': ['lan 10/100/1000 mbps'],
    'nfc': ['nfc (near field communication)'],
    'gps': ['gps'],
    'widi': ['intel wireless display (widi)'],
    '3g_modem': ['modem 3g (wwan)'],
    '4g_modem': ['modem 4g (lte)'],
}

def normalizar_comunicacao(item, mapeamento):
    item_lower = item.lower()  # normaliza maiúscula/minúscula
    for padrao, lista_termos in mapeamento.items():
        if any(termo in item_lower for termo in lista_termos):
            return padrao
    return item_lower

def normalizar_lista(lista_comunicacao, mapeamento):
    normalizada = []
    for item in lista_comunicacao:
        norm_item = normalizar_comunicacao(item, mapeamento)
        if norm_item not in normalizada:
            normalizada.append(norm_item)
    return normalizada

df['communications_norm'] = df['communications_list'].apply(lambda x: normalizar_lista(x, mapping_comunications))
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

mlb = MultiLabelBinarizer()
df_encoded = mlb.fit_transform(df['communications_norm'])
df_comunicacao_encoded = pd.DataFrame(df_encoded, columns=mlb.classes_, index=df.index)

df = pd.concat([df, df_comunicacao_encoded], axis=1)
```
